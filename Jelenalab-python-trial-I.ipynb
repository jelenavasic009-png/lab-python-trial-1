## Problem 4: Harder Loops

**Loop 5:** Below you have a secret message to decode. To decode this message, remove all bad letters from the string. Create a new string called `decoded_message` that is the full message output.

```python
# Napisati petlju (loop) koja prolazi kroz string, izbacuje iz njega "loša slova" (bad letters) i tako spaja preostala slova u novu poruku pod nazivom decoded_message.
# Prvo moramo proći kroz tri glavna koraka:
  #Definisati "loša slova" koja želimo ukloniti (npr. u obliku liste ili drugog stringa).
  #Pokrenuti prazan string decoded_message = "" u koji ćemo dodavati samo dobra slova.
  #Proći kroz tajnu poruku slovo po slovo (pomoću for petlje) i provjeriti nalazi li se trenutno slovo među lošim slovima.
  #Ako se ne nalazi, dodajemo ga u našu novu poruku.
  #Primjer koda u PythonuPošto u tvom zadatku nedostaje tačan tekst tajne poruke i popis loših slova:

  # 1. Početni podaci (tajna poruka i loša slova)
secret_message = "XhXeXlXlXoX XwXoXrXlXdX"
bad_letters = ["X"]  # Ili string "X"

# 2. Kreiramo prazan string za rezultat
decoded_message = ""

# 3. Prolazimo petljom kroz svako slovo u tajnoj poruci
for letter in secret_message:
    # Ako slovo NIJE među lošim slovima, dodajemo ga u rezultat
    if letter not in bad_letters:
        decoded_message += letter

# Ispisujemo konačan rezultat
print(decoded_message)
```

**Loop 6:** Again, we have a coded message, this time as a **Caeser cipher**. A Caeser cipher is one in which all letters are replaced with the letter 3 letters to the right. So, A becomes D, B becomes E, etc. To help, I've provided the decoding cipher. Use this cipher to decode the string. Call the decoded string `decoded_message` and print it to the screen.

_Hint:_ Characters in the message that are not in the cipher are to be left as-is. You can do this by using `.get()` instead of using `[]`s. For example:

* `cipher['?']` gives an error
* `cipher.get('?', '?')` returns `?` since `?` was not found in `cipher`

```python
# Ovdje imamo Cezarovu sifru (Caesar cipher), gdje je svako slovo u poruci pomjereno za 3 mjesta udesno.
# Da bi dešifrovali poruku, kod mora da prođe kroz šifrovani string slovo po slovo, zamijeni svako šifrovano slovo njegovim originalnim slovom pomoću datog rječnika (cipher) i spoji ih u novi string decoded_message.
# Glavni trik u ovom zadatku je rukovanje znakovima koji nijesu slova (kao što su razmaci, znakovi interpunkcije !, ?, tačke ili brojevi).
# Oni se ne nalaze u rječniku šifre i moraju ostati nepromijenjeni.
# Savjet iz zadatka objasnjava da ako pokušavamo da potražimo znak koji ne postoji u rečniku koristeći uglaste zagrade cipher['?'], Python će prijaviti grešku (KeyError) i program će prestati da radi.
# Zato se koristi metoda .get(ključ, podrazumijevana_vrijednost):cipher.get(letter, letter) radi sljedeće: potraži slovo u rječniku.
# Ako ga nađe, vrati njegov prevod. Ako ga ne nađe (jer je to npr. razmak ili upitnik), vrati to isto slovo/znak nepromijenjeno.



cipher = {
    'A': 'D', 'B': 'E', 'C': 'F', 'D': 'G', 'E': 'H',
    'F': 'I', 'G': 'J', 'H': 'K', 'I': 'L', 'J': 'M',
    'K': 'N', 'L': 'O', 'M': 'P', 'N': 'Q', 'O': 'R',
    'P': 'S', 'Q': 'T', 'R': 'U', 'S': 'V', 'T': 'W',
    'U': 'X', 'V': 'Y', 'W': 'Z', 'X': 'A', 'Y': 'B',
    'Z': 'C'}


# decoded_message = "" — Počinje sa praznim stringom.
# for letter in coded_message: — Petlja uzima jedan po jedan znak iz šifrovane poruke.
# cipher.get(letter, letter) — Ako petlja naiđe na 'H', rječnik će vratiti 'E'.
#Ako naiđe na '!' (koji nije u rječniku), metoda .get() će bez greške vratiti '!'.
# decoded_message += decoded_letter — Svaki dobijeni znak se nadovezuje na kraj novog stringa.
```

```python
coded_message = """F AL JV EXFO QLPP
ZEBZH JV KXFIP
YXYV ELT VLR CBBIFK'?
CBBIFKD DLLA XP EBII"""

# Pokrećemo prazan string za dekodiranu poruku
decoded_message = ""

# Prolazimo kroz šifrovanu poruku slovo po slovo
for letter in coded_message:
    # Koristimo .get() da izbjegnemo grešku za znakove koji nijesu u rječniku
    # Ako znak ne postoji u rječniku (npr. razmak ili znak pitanja), ostaje kakav jeste
    decoded_message += cipher.get(letter, letter)

# Štampamo dekodiranu poruku na ekran
print(decoded_message)
```
